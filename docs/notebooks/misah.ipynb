## Orientations

```python
Axis = misah.orientations.orien3d.Axis
GeolPlane = misah.orientations.orien3d.GeolPlane
```

```python
gplane = GeolPlane(90, 45)
```

```python
naxis = gplane.normal_axis()
```

```python
naxis
```

```python
naxis.trend
```

```python
naxis.plunge
```

```python
gplane = GeolPlane(315, 10)
```

```python
gplane.azimuth
```

```python
naxis.trend
```
