## Using misah

```python
from misah import geometry
```

```python
from misah.geometry.space2d import Point2D
```

```python
p = Point2D(3, 4)
```

```python
p2 = Point2D(7, 8)
```

```python
p.distance(p2)
```

```python
p.x
```

```python
p.y
```

```python
p.x
```

```python
from misah.geom3d import Point3D
```

```python
P1 = Point3D(1,2,3)
```

```python
P2 = Point3D(3,4,5)
```

```python
P1.distance(P2)
```

```python
P1.x, P1.y, P1.z
```

```python
P1.delta_z(P2)
```

```python
P1.delta_x(P2)
```
